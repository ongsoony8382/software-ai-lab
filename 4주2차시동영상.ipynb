```python
# 2부터 9사이의 정수를 입력받아 구구단 출력

num = int(input("2부터 9 사이의 정수를 입력: "))

i = 1
while i < 10 :
    print(f"{num} x {i} = {num*i}")
    i = i+1
```

```python
# 학생 10명의 성적을 입력받은 다음, 평균을 출력

i = 1
sum = 0
while i < 11:
    s = int(input(f"학생{i} 성적 입력: "))
    sum += s 
    i += 1

print(f"평균은 { sum/10 }점입니다.")
```

```python
# 학생의 수를 정확히 모른다고 가정하자. -1이 입력될 때까지 학생들의 성적을 입력받은 다음, 그때까지 입력된 성적의 평균 출력

i = 0
total = 0
while True:
    s = int(input("학생의 성적을 입력하세요: "))
    if s == -1:
        break
    
    i += 1 
    total += s

print(f"학생들의 평균 성적은 {total / i}입니다.")
```

```python
# 양의 정수 n을 입력받은 다음, 소수이면 True를, 소수가 아니면 False를 출력하라. 
# 소수: 1과 자기 자신을 제외하고는 나누어 떨어지지 않는 정수 (약수가 딱 2개인 수)

n = int(input("양의 정수를 입력: "))

if n < 2: # 1은 소수가 아님.
    print("False") 
else:
    i = 2
    while i < n:
        if n % i == 0:
            print("False")
            break
        i += 1
    else:
        print("True")
        
    

```
