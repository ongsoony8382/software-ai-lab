```python
# 정수를 입력받아 구구단 출력
n = int(input("2와 9사이의 정수 입력: "))

if n >=2 and n <= 9:
  print(f"{n}단")
  for i in range(1, 10):
    print(f"{n}*{i}={n*i}")
else:
  print("2에서 9 사이의 값을 입력하세요.")
```

```python
# 피보나치 수열은 k번째 항의 값과 k+1 번째 항의 값을 더한 값이 k+2번째 항의 값이 되는 수열을 의미한다. 단, 첫 번째, 두 번째 항의 값은 모두 1이다.
# 1, 1, 2, 3, 5, 8, 13, ...

# 사용자로부터 항의 번호를 입력받은 후, 그 항의 값을 출력하는 프로그램을 작성해보자. 예를들면 사용자가 4를 입력하면 3을 출력하고, 7을 입력하면 13을 출력한다.

a = []
n = int(input("항 번호를 입력하세요: "))
a.append(1)
a.append(1)

if (n > 2):
  for i in range(2, n):
    a.append(0)
    a[i] = a[i-1] + a[i-2]
  print(a[n-1])
else :
  print(1)


```
