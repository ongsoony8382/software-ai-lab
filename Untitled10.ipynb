## 과제 8: 22212055 김수연

```python
import math

def sphere_Volume(radius):
    volume = (4.0/3.0) * math.pi * radius * radius * radius
    return volume;

radius = float(input("구의 반지름을 입력하세요:"))
sphere_Volume(radius)
```
