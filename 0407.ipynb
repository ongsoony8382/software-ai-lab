```python
# 중첩 반복문
x = 1
while x < 4:
    y = 1
    while y < 4 :
        print(x, ",", y)
        y += 1
    x += 1
print("반복문이 끝났습니다.")
```

```python
# while문을 이용한 중첩 반복문
# -1이 입력될 때까지 여러 개의 양의 정수들을 입력받은 다음 각 정수의 factorial을 계산하라.

while True:
    n = int(input("양의 정수를 입력(종료는 -1)"))
    if (n == -1):
        break
    f = 1
    for x in range(1, n+1):
        f = f * x
    print(f"{n}!={f}")
```

```python
A = [23, 15, 2, 14, 14, 16, 20, 14]
B = [2, 48, 15, 14, 26, 32, 47, 14]

# 해결 방법 : 중첩 반복문, 리스트 함축(list comprehension), set 함수
```

```python
# 방법 1. for 문으로 구현

C = []
for item in A:
    if item in B:
        C.append(item)
print(C)
```

```python
# 방법 2. 리스트 함축 (list comprehension)을 이용하는 방법

c = [x for x in A if x in B]
print(c)
```

```python
# 방법 3. set 연산을 이용하는 방법 ( 중복 제거, 오름차순)

C = list(set(A).intersection(B))
print(C)
```
