```python
# 반지름이 3인 원의 면적과 구의 부피 출력

r = 3
pi = 3.14159
area = pi * r ** 2
volume = 4 / 3 * pi * r ** 3
print(f"반지름이 {r}인 원의 면적은 {area}, 부피는 {volume:.2f}입니다.")
```
