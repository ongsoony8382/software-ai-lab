```python
# Turtle 그래픽스
# 명령어로 거북이(또는 펜)를 이동시켜 원하는 도형을 그리는 도구

import turtle #turtle 그래픽스를 사용하기 위하여 모듈을 가져오기 
t = turtle.Turtle()
t.shape("turtle")
t.color("red")
t.forward(100)
t.left(90)
t.forward(100)
t.left(90)
t.forward(100)
t.left(90)
t.forward(100)
turtle.mainloop()
turtle.bye()
```

```python
# ▪ 우리 집 야식으로 피자와 치킨, 그리고 케이크를 주문하려고 한다.
# ▪ 피자 가격은 15,000원이며, 치킨은 7,000원, 그리고 케이크는 3,000원이라고 가정하자.
# ▪ 가족 모두 피자를 가장 좋아하고, 다음으로 치킨을 좋아한다. 케이크에 대한 선호도는 가장 낮다.
# ▪ 따라서 예산이 허락하는 한 피자를 많이 주문하고, 남는 돈으로 치킨을 최대한 주문하자. 나머지 잔돈으로 케이크를 주문하려고 한다.
# ▪ 예산을 입력받아, 피자와 치킨, 케이크를 몇 개씩 주문할 수 있는지를 계산하여 출력하라.
# ▪ 예로 예산이 30,000원이면 피자를 두 판 주문하는 것이 최선이고, 40,000원이면 피자 두 판에 치킨 한
# 마리, 케이크 하나를 주문한다. 예산이 50,000원이면 피자 세 판에 케아크 하나를 주문하고 2,000원을 거스름돈으로 받는다.
```

```python
budget = int(input("예산을 입력하세요: "))
pizza = 15000
chicken = 7000
cake = 3000

num_pizza = budget//pizza # 구매할 피자 개수 
budget = budget % pizza # 피자를 사고 남은 금액
num_chicken = budget // chicken # 구매할 치킨 수 
budget = budget % chicken # 치킨을 사고 남은 금액
num_cake = budget // cake # 구매할 케이크 수 
change = budget % cake # 케이크까지 구매 후 남은 금액  

print(f"피자 {num_pizza}판, 치킨 {num_chicken}마리,", end=" ")
print(f"케이크 {num_cake}접시, 잔돈 = {change}원")
```

```python
budget = int(input("예산을 입력하세요: "))
num_pizza = budget // 15000 
num_chi = (budget%15000)//7000
num_cake = (budget - (15000*num_pizza) - (7000*num_chi))//3000 
print(f"예산 {budget}으로 피자 {num_pizza}판, 치킨 {num_chi}마리, 케익 {num_cake}접시 가능")
```
